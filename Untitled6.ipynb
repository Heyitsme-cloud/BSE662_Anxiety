```python
import os
import warnings
import pandas as pd
import numpy as np
import mne
from fooof import FOOOF

warnings.filterwarnings("ignore")  # Ignore warnings

# PATHS
trial_csv = "/content/EEG_trial_segmentation - 47131_pre.csv"

# Load trial metadata
df = pd.read_csv(trial_csv)

# Frontal channel pairs (adjust based on your montage)
left_frontal = ['FP1', 'F3', 'F7']
right_frontal = ['FP2', 'F4', 'F8']

def extract_features(raw, tmin, tmax):
    try:
        # Crop to trial
        raw_crop = raw.copy().crop(tmin=tmin, tmax=tmax)
        raw_crop.pick("eeg")

        # Compute PSD (Welch)
        psd_obj = raw_crop.compute_psd(method='welch', fmin=1, fmax=50)
        freqs = psd_obj.freqs
        psd_data = psd_obj.get_data()

        # PSD averaged over all channels
        psd_avg = np.mean(psd_data, axis=0)

        # === 1. Beta power (13–30 Hz) ===
        beta_mask = (freqs >= 13) & (freqs <= 30)
        beta_powers = np.mean(psd_data[:, beta_mask], axis=1)
        beta_avg = np.mean(beta_powers)

        # === 2. Frontal Asymmetry ===
        ch_names = raw_crop.info['ch_names']
        left_beta, right_beta = [], []

        for ch in left_frontal:
            if ch in ch_names:
                idx = ch_names.index(ch)
                left_beta.append(np.mean(psd_data[idx, beta_mask]))

        for ch in right_frontal:
            if ch in ch_names:
                idx = ch_names.index(ch)
                right_beta.append(np.mean(psd_data[idx, beta_mask]))

        if left_beta and right_beta:
            frontal_asymmetry = np.log(np.mean(left_beta)) - np.log(np.mean(right_beta))
        else:
            frontal_asymmetry = None

        # === 3. Aperiodic exponent via FOOOF ===
        fm = FOOOF(peak_width_limits=[1, 12], max_n_peaks=6)
        fm.fit(freqs, psd_avg)
        exponent = fm.aperiodic_params_[1] if fm.aperiodic_params_ is not None else None

        return beta_avg, psd_avg.tolist(), freqs.tolist(), exponent, frontal_asymmetry

    except Exception as e:
        print(f"Trial skipped due to error: {e}")
        return None, None, None, None, None

# Process trials per file
results = []
for file_name, file_df in df.groupby('file'):
    fif_path = "/content/47131_cleaned.fif"

    if not os.path.isfile(fif_path):
        print(f"File not found: {fif_path}")
        continue

    print(f"Processing: {file_name}")
    try:
        raw = mne.io.read_raw_fif(fif_path, preload=True, verbose=False)
    except Exception as e:
        print(f"Could not load {file_name}: {e}")
        continue

    for _, row in file_df.iterrows():
        tmin = row['trial_start_sec']
        tmax = tmin + row['duration_sec']
        trial_idx = row['trial_index']

        beta, psd, freqs, exponent, asym = extract_features(raw, tmin, tmax)

        results.append({
            "file": file_name,
            "trial_index": trial_idx,
            "beta_power_avg": beta,
            "aperiodic_exponent": exponent,
            "frontal_asymmetry": asym
        })

# Show results
results_df = pd.DataFrame(results)
results_df.head(10)
```
